#Clean The Dataset

```python
df.signup_date = pd.to_datetime(df.signup_date, errors='coerce')
```

```python
df.plan_selected = df.plan_selected.str.strip().str.title()
df.marketing_opt_in = df.marketing_opt_in.str.strip().str.title()
df.source = df.source.str.strip().str.title()
df.region = df.region.str.strip().str.title()
df.gender = df.gender.str.strip().str.title()
```

```python
df.head(10)
```

```python
df.duplicated('customer_id').sum()
```

```python
df.isnull().sum()
```

```python
#replace rows with missing region value
df["region"] = df["region"].fillna("Unknown")
```

```python
#remove rows with missing column id value
df = df[df["customer_id"].notna()]
```

```python
#replace rows with missing plan selected value
df.loc[:,"plan_selected"] = df["plan_selected"].fillna("Not Selected")
```

```python
#remove rows with missing name value
df = df[df['name'].notna()]
```

```python
#marketing_opt_in with missing value since the value is "Yes" or "No" if they didn't select it, it should be No
df["marketing_opt_in"] = df["marketing_opt_in"].fillna("No")
```

```python
# age convert invalid ages to missing
df["age"] = pd.to_numeric(df["age"], errors="coerce")
```

```python
# age convert invalid ages to missing
df["gender"] = df["gender"].fillna("Unknown")
```
