#Feature Engineering

```python
#1. Revenue 
total_revenue = (
    df_mergedSPC['quantity']
    * df_mergedSPC['unit_price']
    * (1 - df_mergedSPC['discount_applied'])
).sum()

print(total_revenue)
```

```python
#2. Order week
df_mergedSPC['order_week'] = df_mergedSPC['order_date'].dt.isocalendar().week
df_mergedSPC.head(5)
```

```python
#3. Price Band
df_mergedSPC['price_band'] = pd.cut(df_mergedSPC['unit_price'], bins=[-1, 15, 30, float('inf')], labels=['Low', 'Medium', 'High'])
df_mergedSPC.head(5)
```

```python
#4. Days to Order 
df_mergedSPC['days_to_order'] = (df_mergedSPC['order_date'] - df_mergedSPC['launch_date']).dt.days
df_mergedSPC.head(5)
```

```python
#5. Email Domain
df_mergedSPC['email_domain'] = df_mergedSPC['email'].str.split('@').str[1]
df_mergedSPC.head(5)
```

```python
#6. Late delivery flag 
df_mergedSPC['is_late'] = (df_mergedSPC['delivery_status'] == 'Delayed')
df_mergedSPC.head(5)
```

#Create Table Summary 

```python
#1. Weekly revenue trend by region 
df_mergedSPC['revenue'] = df_mergedSPC['quantity'] * df_mergedSPC['unit_price'] * (1 - df_mergedSPC['discount_applied'])
weekly_revenue = (
    df_mergedSPC.groupby([df_mergedSPC['region_x'], df_mergedSPC['order_date'].dt.isocalendar().week])
    ['revenue'].sum().reset_index(name='weekly_revenue')
    )
weekly_revenue.rename(columns={'week': 'week_number'}, inplace=True)
df_mergedSPC.head(5)
```

```python
#2. Product Category Performance 
product_category_perf = (
    df_mergedSPC.groupby('category')
    .agg(
        revenue_sum=('revenue', 'sum'),
        quantity_sum=('quantity', 'sum'),
        discount_applied_sum=('discount_applied', 'sum')
      )
    .reset_index()
)

print(product_category_perf)
```

```python

pivot_table = product_category_perf.pivot(
    index='discount_applied_sum',
    columns='category',
    values='quantity_sum'
)
print(pivot_table)
```

```python
#3. Customer Behavior
customer_behavior = (
    df_mergedSPC.groupby([df_mergedSPC['loyalty_tier'],df_mergedSPC['signup_date'].dt.to_period('M')])
    .agg(
        order_count = ('order_id', 'count'),
        revenue_sum = ('revenue','sum')
    )
    .reset_index()
)

print(customer_behavior)
```

```python
customer_behavior.describe(include='all')
```

```python
#4. Delivery Performance
Delivery_Performance = (
    df_mergedSPC.groupby(['region_x', 'price_band'])
    .agg(
        order_count=('order_id', 'count'),
        delayed_count=('delivery_status', lambda x: (x == 'Delayed').sum()),
        delivered_count=('delivery_status', lambda x: (x == 'Delivered').sum())
    )
    .reset_index()
)

print(Delivery_Performance)
```

```python
#5. Payment Preferences
Payment_Preferences = (
    df_mergedSPC.groupby(['loyalty_tier', 'payment_method'])
    .agg(
        transaction_count=('order_id', 'count')
    )
    .reset_index()
)

print(Payment_Preferences)
```

```python
#1. Weekly revenue trend by region 
plt.figure(figsize=(12,6))
sns.lineplot(data=weekly_revenue, x='week_number', y='weekly_revenue', hue='region_x', marker='o')
plt.title("Weekly Revenue Trends by Region")
plt.xlabel("Week Number")
plt.ylabel("Revenue")
plt.show()
```

```python
#2. Top five product category by revenue 
plt.figure(figsize=(6,3))
sns.barplot(data=product_category_perf, x='category', y='revenue_sum', hue='category')
plt.title("Top five product category by revenue")
plt.xlabel("Category")
plt.ylabel("Revenue")
plt.show()
```

```python
#3. Quantity versus discount by category 
plt.figure(figsize=(12,6))
sns.boxplot(data=product_category_perf, x='quantity_sum', y='discount_applied_sum', hue='category',palette='pastel', width=0.6, fliersize=5)
plt.title("Quantity versus Discount by category")
plt.xlabel("Quantity")
plt.ylabel("Discount Applied")
plt.show()
```

```python
#4. Correlation between revenue, discount and quantity
cols = ['revenue', 'discount_applied', 'quantity']
corr_matrix = df_mergedSPC[cols].corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap='coolwarm', cbar=True)
plt.title("Correlation between Revenue, Discount Applied, and Quantity")
plt.show()
```

```python
#Which product categories drive the most revenue, and in which regions?
top_categories = (
    df_mergedSPC.groupby("category")["revenue"]
    .sum()
    .sort_values(ascending=False)
)
print(top_categories)

top_regions = (
    df_mergedSPC.groupby("region_x")["revenue"]
    .sum()
    .sort_values(ascending=False)
)
print(top_regions)
```

```python
revenue_by_tier = (
    df_mergedSPC.groupby("loyalty_tier")["revenue"]
    .sum()
    .reset_index()
    .sort_values("revenue", ascending=False)
)
print(revenue_by_tier)
```

```python
late_by_region = (
    df_mergedSPC.groupby("region_x")["is_late"]
    .value_counts()
    .unstack(fill_value=0)
    .reset_index()
)
print(late_by_region)
```

```python
# Extract signup month
df_mergedSPC['signup_month'] = df_mergedSPC['signup_date'].dt.to_period('M')

# Aggregate by signup month: total orders and average revenue
signup_summary = df_mergedSPC.groupby('signup_month').agg(
    total_orders=('order_id', 'count'),
    avg_revenue=('revenue', 'mean')
).reset_index()

print(signup_summary)
```
